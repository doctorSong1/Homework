### 1 데이터 합치기

```python
# 1-1 두 데이터프레임을 합치기

combined_cars = pd.concat([df1, df2])
combined_cars.info()
combined_cars.head(10)
```

```python
# 1-2 카테고리형(Engine, Emission Class) → 숫자로 바꾸기 

# Engine: "1.6L" -> 1.6
df1["Engine"] = df1["Engine"].astype(str).str.replace("L", "", regex=False)
df1["Engine"] = pd.to_numeric(df1["Engine"], errors="coerce")

# Emission Class: "Euro 6" -> 6
df1["Emission Class"] = df1["Emission Class"].astype(str).str.extract(r"(\d+)")
df1["Emission Class"] = pd.to_numeric(df1["Emission Class"], errors="coerce")
df1
```

```python
# 1-3 Service history 결측치 채우기

df1["Service history"] = df1["Service history"].fillna("Unknown")

# 확인
df1.isna().sum()
```

```python
# 1-3 결측치 처리

# Service history 결측치 'Unknown'으로 채우기
df1["Service history"] = df1["Service history"].fillna("Unknown")

# 결측값 1411인 'Previous Owners' column을 삭제
df1 = df1.drop('Previous Owners', axis =1)

# 확인
df1.isna().sum() 
```

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# 1) 숫자 컬럼만 가져오기
X = df1.select_dtypes(include=["number"]).copy()

# 2) inf -> NaN
X = X.replace([np.inf, -np.inf], np.nan)

# 3) NaN 채우기 (행이 0개 되는 것 방지)
X = X.fillna(X.mean())

# 4) 혹시 평균으로도 못 채운 컬럼(전부 NaN)이 있으면 그 컬럼 삭제
X = X.dropna(axis=1)

print("X shape:", X.shape)   # 여기서 (0, ?)면 df1 자체가 비어있는 상태

# 5) 표준화 + PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.70)
X_pca = pca.fit_transform(X_scaled)

print("주성분 개수:", pca.n_components_)
print("설명력 합:", pca.explained_variance_ratio_.sum())
```
