```python
# 시간 컬럼 변환 및 정렬

df['trans_date_trans_time'] = pd.to_datetime(df['trans_date_trans_time'], errors='coerce')

# 결측/이상치 확인
print("trans_date_trans_time null:", df['trans_date_trans_time'].isna().sum())

# 시간 기준 정렬 (필수 사항)
df = df.sort_values('trans_date_trans_time').reset_index(drop=True)

# 타깃 확인
print(df['is_fraud'].value_counts(dropna=False))
print()
print("fraud rate:", df['is_fraud'].mean())
```

```python
# 누수 후보 컬럼 점검 

leak_keywords = [
    "fraud", "chargeback", "cbk", "dispute", "investigation", "report",
    "label", "target", "outcome", "result", "post", "after"
]

suspects = [c for c in df.columns if any(k in c.lower() for k in leak_keywords)]
print("Suspicious columns (name-based):", suspects)
```

```python
# Train - Test 분리 8: 2

# 시간 기준 hold-out split (최근 20%를 test)
split_idx = int(len(df) * 0.8)

train_df = df.iloc[:split_idx].copy()
test_df  = df.iloc[split_idx:].copy()

print("train period:", train_df['trans_date_trans_time'].min(), "->", train_df['trans_date_trans_time'].max())
print("test  period:", test_df['trans_date_trans_time'].min(), "->", test_df['trans_date_trans_time'].max())
print()
print("train fraud rate:", train_df['is_fraud'].mean())
print("test  fraud rate:", test_df['is_fraud'].mean())
```

```python
# 즉시 누수 차단 

TARGET = "is_fraud"
TIME_COL = "trans_date_trans_time"

# 제외 권장(암기/의미 약함/식별자 성격)
drop_cols = [TARGET, TIME_COL, "trans_num"]  # trans_num이 없으면 자동으로 무시되게 아래에서 처리

drop_cols = [c for c in drop_cols if c in df.columns]

X_train = train_df.drop(columns=drop_cols)
y_train = train_df[TARGET].astype(int)

X_test  = test_df.drop(columns=drop_cols)
y_test  = test_df[TARGET].astype(int)

print("X_train shape:", X_train.shape, "X_test shape:", X_test.shape)
```

```python
# 안전한 베이스라인 모델 (Pipeline로 누수 방지)
# 범주형: OneHotEncoder
# 수치형: 결측 처리
# 모델: LogisticRegression(class_weight='balanced') → 불균형 기본 대응
```

```python
# 베이스 라인 모델 
# PR-AUC 값이 중요함

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score, roc_auc_score, classification_report

# 컬럼 타입 분리
cat_cols = X_train.select_dtypes(include=["object"]).columns.tolist()
num_cols = [c for c in X_train.columns if c not in cat_cols]

preprocess = ColumnTransformer(
    transformers=[
        ("num", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="median")),
        ]), num_cols),
        ("cat", Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ohe", OneHotEncoder(handle_unknown="ignore")),
        ]), cat_cols),
    ],
    remainder="drop"
)

model = Pipeline(steps=[
    ("prep", preprocess),
    ("clf", LogisticRegression(max_iter=200, class_weight="balanced", n_jobs=None))
])

model.fit(X_train, y_train)

proba = model.predict_proba(X_test)[:, 1]

print("PR-AUC (Average Precision):", average_precision_score(y_test, proba))
print("ROC-AUC:", roc_auc_score(y_test, proba))
```

```python
# 임계값Threadhold 조정, 
# (놓치지 않기 위해 Recall 값을 기본값 0.5 보다 낮춤) 

from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_test, proba)

# 예: recall이 0.80 이상이 되는 가장 낮은 threshold 찾기
target_recall = 0.80
idx = np.where(recall >= target_recall)[0]
if len(idx) > 0:
    best_idx = idx[-1]  # recall 조건 만족하는 지점 중 threshold가 상대적으로 높은 쪽(precision 조금이라도 확보)
    th = thresholds[best_idx-1] if best_idx > 0 else 0.0
else:
    th = 0.5

pred = (proba >= th).astype(int)
print("chosen threshold:", th)
print(classification_report(y_test, pred, digits=4))
```

```python
# =========================
# Step A) Basic check & sort
# =========================

# 2) 핵심 컬럼명 고정
TIME_COL = "trans_date_trans_time"
TARGET   = "is_fraud"

# 3) 필수 컬럼 존재 확인
required = [TIME_COL, TARGET]
missing = [c for c in required if c not in df.columns]
if missing:
    raise KeyError(f"필수 컬럼이 없습니다: {missing}\n현재 컬럼 예시: {df.columns.tolist()[:20]}")

# 4) 시간 컬럼 datetime 변환
df[TIME_COL] = pd.to_datetime(df[TIME_COL], errors="coerce")

# 5) 시간 결측 확인 (결측이 많으면 다음 단계 전에 처리 필요)
null_time = df[TIME_COL].isna().sum()
print("TIME null count:", null_time)

# 6) 시간 기준 정렬 (누수 방지의 출발점)
df = df.sort_values(TIME_COL).reset_index(drop=True)

# 7) 타깃 분포 확인
print("\nTARGET distribution:")
print(df[TARGET].value_counts(dropna=False))
print("fraud rate:", df[TARGET].mean())

# 8) 최소 확인 출력
display(df[[TIME_COL, TARGET]].head(5))
display(df[[TIME_COL, TARGET]].tail(5))
print("\nshape:", df.shape)
```

```python
# 2) 컬럼명 공백 제거 (필수)
df.columns = df.columns.str.strip()

# 3) 컬럼명 고정
TIME_COL = "trans_date_trans_time"
TARGET   = "is_fraud"
KEY_COL  = "cc_num"
AMT_COL  = "amt"

# 4) 필수 컬럼 체크
required = [TIME_COL, TARGET, KEY_COL, AMT_COL]
missing = [c for c in required if c not in df.columns]
if missing:
    raise KeyError(f"필수 컬럼이 없습니다: {missing}\n현재 컬럼(일부): {df.columns.tolist()[:30]}")

# 5) 시간 변환 + 정렬
df[TIME_COL] = pd.to_datetime(df[TIME_COL], errors="coerce")
if df[TIME_COL].isna().sum() != 0:
    raise ValueError("TIME_COL에 datetime 변환 실패(NaT)가 있습니다. 먼저 확인이 필요합니다.")

df = df.sort_values(TIME_COL).reset_index(drop=True)

# 6) time-based split
split_idx = int(len(df) * 0.8)
train_df = df.iloc[:split_idx].copy()
test_df  = df.iloc[split_idx:].copy()

# 7) 확인
print("✅ OK")
print("Train columns has TIME_COL? ->", TIME_COL in train_df.columns)
print("Test  columns has TIME_COL? ->", TIME_COL in test_df.columns)
print("Train period:", train_df[TIME_COL].min(), "->", train_df[TIME_COL].max())
print("Test  period:", test_df[TIME_COL].min(), "->", test_df[TIME_COL].max())
print("Train fraud rate:", train_df[TARGET].mean())
print("Test  fraud rate:", test_df[TARGET].mean())
```

```python
# =========================
# Step C-1) Rolling count (1h) - ONE feature only
# =========================

# 0) 안전하게 컬럼명 strip (혹시 모를 공백 방지)
train_df.columns = train_df.columns.str.strip()
test_df.columns  = test_df.columns.str.strip()

# 1) 고객+시간 정렬 (필수)
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
test_df  = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)

# 2) Train: 시간 인덱스 기반 rolling (현재 거래 제외)
train_df = train_df.set_index(TIME_COL)
train_df["cc_txn_cnt_1h"] = (
    train_df[AMT_COL]
    .groupby(train_df[KEY_COL])
    .rolling("1h", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)
train_df = train_df.reset_index()

# 3) Test: 동일 로직 (Train과 섞지 않음)
test_df = test_df.set_index(TIME_COL)
test_df["cc_txn_cnt_1h"] = (
    test_df[AMT_COL]
    .groupby(test_df[KEY_COL])
    .rolling("1h", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)
test_df = test_df.reset_index()

# 4) 검증(간단): 값 분포 확인
print("✅ created: cc_txn_cnt_1h")
print("train cc_txn_cnt_1h describe:\n", train_df["cc_txn_cnt_1h"].describe())
print("\ntest  cc_txn_cnt_1h describe:\n", test_df["cc_txn_cnt_1h"].describe())

# 상위 몇 개 확인
display(train_df[[TIME_COL, KEY_COL, AMT_COL, "cc_txn_cnt_1h"]].head(10))
```

```python
# =========================
# Step C-2) Rolling sum (1h) - ONE feature only
# =========================

# 1) Train
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
train_df = train_df.set_index(TIME_COL)

train_df["cc_amt_sum_1h"] = (
    train_df[AMT_COL]
    .groupby(train_df[KEY_COL])
    .rolling("1h", closed="left")
    .sum()
    .reset_index(level=0, drop=True)
).fillna(0)

train_df = train_df.reset_index()

# 2) Test
test_df = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
test_df = test_df.set_index(TIME_COL)

test_df["cc_amt_sum_1h"] = (
    test_df[AMT_COL]
    .groupby(test_df[KEY_COL])
    .rolling("1h", closed="left")
    .sum()
    .reset_index(level=0, drop=True)
).fillna(0)

test_df = test_df.reset_index()

# 3) 확인
print("✅ created: cc_amt_sum_1h")
print("train cc_amt_sum_1h describe:\n", train_df["cc_amt_sum_1h"].describe())
print("\ntest  cc_amt_sum_1h describe:\n", test_df["cc_amt_sum_1h"].describe())

display(train_df[[TIME_COL, KEY_COL, AMT_COL, "cc_txn_cnt_1h", "cc_amt_sum_1h"]].head(10))
```

```python
# =========================
# Step C-3) Rolling count (1d) - ONE feature only
# =========================

# Train
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
train_df = train_df.set_index(TIME_COL)

train_df["cc_txn_cnt_1d"] = (
    train_df[AMT_COL]
    .groupby(train_df[KEY_COL])
    .rolling("1d", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)

train_df = train_df.reset_index()

# Test
test_df = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
test_df = test_df.set_index(TIME_COL)

test_df["cc_txn_cnt_1d"] = (
    test_df[AMT_COL]
    .groupby(test_df[KEY_COL])
    .rolling("1d", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)

test_df = test_df.reset_index()

# 확인
print("✅ created: cc_txn_cnt_1d")
print("train cc_txn_cnt_1d describe:\n", train_df["cc_txn_cnt_1d"].describe())
print("\ntest  cc_txn_cnt_1d describe:\n", test_df["cc_txn_cnt_1d"].describe())

display(train_df[[TIME_COL, KEY_COL, AMT_COL, "cc_txn_cnt_1h", "cc_txn_cnt_1d"]].head(10))
```

```python
# =========================
# Step C-4) Rolling sum (1d) - ONE feature only
# =========================

# Train
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
train_df = train_df.set_index(TIME_COL)

train_df["cc_amt_sum_1d"] = (
    train_df[AMT_COL]
    .groupby(train_df[KEY_COL])
    .rolling("1d", closed="left")
    .sum()
    .reset_index(level=0, drop=True)
).fillna(0)

train_df = train_df.reset_index()

# Test
test_df = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
test_df = test_df.set_index(TIME_COL)

test_df["cc_amt_sum_1d"] = (
    test_df[AMT_COL]
    .groupby(test_df[KEY_COL])
    .rolling("1d", closed="left")
    .sum()
    .reset_index(level=0, drop=True)
).fillna(0)

test_df = test_df.reset_index()

# 확인
print("✅ created: cc_amt_sum_1d")
print("train cc_amt_sum_1d describe:\n", train_df["cc_amt_sum_1d"].describe())
print("\ntest  cc_amt_sum_1d describe:\n", test_df["cc_amt_sum_1d"].describe())

display(train_df[[TIME_COL, KEY_COL, AMT_COL, "cc_amt_sum_1h", "cc_amt_sum_1d"]].head(10))
```

```python
# =========================
# Step C-5) Rolling count (7d) - ONE feature only
# =========================

# Train
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
train_df = train_df.set_index(TIME_COL)

train_df["cc_txn_cnt_7d"] = (
    train_df[AMT_COL]
    .groupby(train_df[KEY_COL])
    .rolling("7d", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)

train_df = train_df.reset_index()

# Test
test_df = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)
test_df = test_df.set_index(TIME_COL)

test_df["cc_txn_cnt_7d"] = (
    test_df[AMT_COL]
    .groupby(test_df[KEY_COL])
    .rolling("7d", closed="left")
    .count()
    .reset_index(level=0, drop=True)
).fillna(0)

test_df = test_df.reset_index()

# 확인
print("✅ created: cc_txn_cnt_7d")
print("train cc_txn_cnt_7d describe:\n", train_df["cc_txn_cnt_7d"].describe())
print("\ntest  cc_txn_cnt_7d describe:\n", test_df["cc_txn_cnt_7d"].describe())

display(train_df[[TIME_COL, KEY_COL, AMT_COL, "cc_txn_cnt_1d", "cc_txn_cnt_7d"]].head(10))
```

```python
# =========================
# Step C-6) Time since previous transaction (ONE feature)
# =========================

# Train
train_df = train_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)

train_df["since_prev_txn_sec"] = (
    train_df
    .groupby(KEY_COL)[TIME_COL]
    .diff()
    .dt.total_seconds()
)

# 첫 거래는 이전 거래가 없으므로 NaN -> 0 처리
train_df["since_prev_txn_sec"] = train_df["since_prev_txn_sec"].fillna(0)


# Test
test_df = test_df.sort_values([KEY_COL, TIME_COL]).reset_index(drop=True)

test_df["since_prev_txn_sec"] = (
    test_df
    .groupby(KEY_COL)[TIME_COL]
    .diff()
    .dt.total_seconds()
)

test_df["since_prev_txn_sec"] = test_df["since_prev_txn_sec"].fillna(0)


# 확인
print("✅ created: since_prev_txn_sec")

print("\ntrain since_prev_txn_sec describe:\n",
      train_df["since_prev_txn_sec"].describe())

print("\ntest since_prev_txn_sec describe:\n",
      test_df["since_prev_txn_sec"].describe())

display(train_df[[TIME_COL, KEY_COL, "since_prev_txn_sec"]].head(10))
```

```python
# =========================
# Step D-1) Baseline model + PR-AUC
# =========================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import average_precision_score

# 1) Feature / Target 분리
drop_cols = [TARGET, TIME_COL]

X_train = train_df.drop(columns=drop_cols)
y_train = train_df[TARGET].astype(int)

X_test  = test_df.drop(columns=drop_cols)
y_test  = test_df[TARGET].astype(int)

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)

# 2) 수치형/범주형 구분
cat_cols = X_train.select_dtypes(include=["object"]).columns.tolist()
num_cols = [c for c in X_train.columns if c not in cat_cols]

print("num cols:", len(num_cols), "| cat cols:", len(cat_cols))

# 3) 전처리 파이프라인
preprocess = ColumnTransformer(
    transformers=[
        ("num", Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ]), num_cols),

        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("ohe", OneHotEncoder(handle_unknown="ignore"))
        ]), cat_cols),
    ]
)

# 4) Logistic Regression (불균형 대응)
model = Pipeline([
    ("prep", preprocess),
    ("clf", LogisticRegression(max_iter=200, class_weight="balanced"))
])

# 5) 학습
model.fit(X_train, y_train)

# 6) 예측 확률
proba = model.predict_proba(X_test)[:, 1]

# 7) PR-AUC 계산
pr_auc = average_precision_score(y_test, proba)

print("\n✅ Baseline PR-AUC (Average Precision):", pr_auc)
```

```python
# =========================
# Step D-2) RandomForest PR-AUC (numeric features only)
# =========================

from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import average_precision_score

# 1) Feature/Target 분리
drop_cols = [TARGET, TIME_COL]
X_train = train_df.drop(columns=drop_cols)
y_train = train_df[TARGET].astype(int)

X_test  = test_df.drop(columns=drop_cols)
y_test  = test_df[TARGET].astype(int)

# 2) 수치형만 사용 (안전/빠른 비교용)
X_train_num = X_train.select_dtypes(exclude=["object"]).copy()
X_test_num  = X_test.select_dtypes(exclude=["object"]).copy()

print("numeric X_train shape:", X_train_num.shape)
print("numeric X_test  shape:", X_test_num.shape)

# 3) 간단 파이프라인: 결측치 처리 + RF
rf_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("rf", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced_subsample",
        min_samples_leaf=10
    ))
])

# 4) 학습/평가
rf_model.fit(X_train_num, y_train)
proba = rf_model.predict_proba(X_test_num)[:, 1]

pr_auc = average_precision_score(y_test, proba)
print("\n✅ RandomForest PR-AUC (numeric-only):", pr_auc)
```

```python
# =========================
# Step D-3) Feature importance TOP 10
# =========================

# RandomForest 내부 모델 꺼내기
rf_clf = rf_model.named_steps["rf"]

# feature 이름 가져오기
feature_names = X_train_num.columns

# importance DataFrame 만들기
importances = pd.DataFrame({
    "feature": feature_names,
    "importance": rf_clf.feature_importances_
})

# 중요도 내림차순 정렬
importances = importances.sort_values("importance", ascending=False)

print("✅ Feature Importance TOP 10")
display(importances.head(10))
```

```python
# =========================
# Step D-4) Top 1% Detection Rate
# =========================

# 예측 확률 다시 생성
proba = rf_model.predict_proba(X_test_num)[:, 1]

# 테스트 데이터에 붙이기
result = test_df.copy()
result["proba"] = proba

# 사기 전체 개수
total_fraud = result[TARGET].sum()

# Top 1% cutoff
top_pct = 0.01
cutoff = np.quantile(result["proba"], 1 - top_pct)

top_result = result[result["proba"] >= cutoff]

# Top 1% 안에 포함된 fraud 개수
top_fraud = top_result[TARGET].sum()

print("✅ Top 1% 탐지 성능")
print("전체 사기 건수:", int(total_fraud))
print("Top 1% 거래 수:", len(top_result))
print("Top 1% 안의 사기 건수:", int(top_fraud))

print("\n📌 Fraud Recall in Top 1% =", top_fraud / total_fraud)
```

```python
# =========================
# Step V1) PR Curve visualization + mark Top-N points
# =========================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve, average_precision_score

# 1) 예측확률/정답 준비 (rf_model + X_test_num 사용)
proba = rf_model.predict_proba(X_test_num)[:, 1]
y_true = test_df[TARGET].astype(int).values

# 2) PR 커브
precision, recall, thresholds = precision_recall_curve(y_true, proba)
ap = average_precision_score(y_true, proba)

# 3) Top-N 지점(1%, 0.5%, 0.1%)을 PR커브 위에 찍기
#    - Top p%는 threshold = quantile(1-p)
result = test_df.copy()
result["proba"] = proba

points = []
for top_pct in [0.01, 0.005, 0.001]:
    cutoff = np.quantile(result["proba"], 1 - top_pct)
    pred = (result["proba"] >= cutoff).astype(int)

    tp = int(((pred == 1) & (result[TARGET] == 1)).sum())
    fp = int(((pred == 1) & (result[TARGET] == 0)).sum())
    fn = int(((pred == 0) & (result[TARGET] == 1)).sum())

    p = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    r = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    points.append((top_pct, p, r, cutoff, tp, (tp+fp)))

# 4) 그래프
plt.figure(figsize=(7, 5))
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title(f"Precision-Recall Curve (AP={ap:.4f})")

# 랜덤 기준선(=fraud rate)
base_rate = y_true.mean()
plt.axhline(base_rate, linestyle="--")
# Top-N 점 표시
for top_pct, p, r, cutoff, tp, nflag in points:
    plt.scatter([r], [p])
    plt.text(r, p, f" Top {top_pct*100:.1f}%\n P={p:.3f}, R={r:.3f}", fontsize=9)

plt.ylim(0, 1.02)
plt.xlim(0, 1.02)
plt.show()

# 5) 표로도 같이 확인(시각화 해석용)
print("✅ Top-N points summary")
for top_pct, p, r, cutoff, tp, nflag in points:
    print(f"Top {top_pct*100:.1f}% | flagged={nflag:4d} | TP={tp:3d} | Precision={p:.4f} | Recall={r:.4f} | cutoff={cutoff:.6f}")
print("base fraud rate:", base_rate)
```

```python
# =========================
# SHAP-3) Side-by-side Waterfall: Fraud vs Normal (high-risk normal)
# =========================

import shap
import numpy as np
import matplotlib.pyplot as plt
from io import BytesIO
from PIL import Image

# 1) Explainer 준비
explainer = shap.TreeExplainer(rf_top5)

# 2) 테스트셋 예측확률(Top5 모델용)
proba_top5_all = rf_top5.predict_proba(X_test_top5)[:, 1]

# 3) Fraud 1건 선택 (첫 번째 사기)
fraud_idx = test_df[test_df[TARGET] == 1].index[0]
x_fraud = X_test_top5.loc[fraud_idx:fraud_idx]
p_fraud = float(proba_top5_all[list(X_test_top5.index).index(fraud_idx)])

# 4) Normal 1건 선택: 정상 중 "가장 위험하게 예측된" 케이스
normal_candidates = test_df[test_df[TARGET] == 0].index
normal_scores = pd.Series(proba_top5_all, index=X_test_top5.index).loc[normal_candidates]
normal_idx = int(normal_scores.sort_values(ascending=False).index[0])

x_normal = X_test_top5.loc[normal_idx:normal_idx]
p_normal = float(proba_top5_all[list(X_test_top5.index).index(normal_idx)])

print("Fraud idx:", fraud_idx, "| proba:", p_fraud)
print("Normal idx:", normal_idx, "| proba:", p_normal)
print("\nFraud case values:")
display(x_fraud)
print("\nNormal case values:")
display(x_normal)

# 5) Waterfall을 "이미지로" 만든 다음, 한 Figure에 배치하기 위한 함수
def waterfall_to_image(x_case, title):
    sv = explainer(x_case)
    sv_pos = sv[..., 1]  # class=1(사기)만 선택

    shap.plots.waterfall(sv_pos[0], max_display=10, show=False)
    plt.title(title)
    plt.tight_layout()

    buf = BytesIO()
    plt.gcf().savefig(buf, format="png", dpi=160, bbox_inches="tight")
    plt.close(plt.gcf())
    buf.seek(0)
    return np.array(Image.open(buf))

# 6) 두 이미지 생성
img_fraud = waterfall_to_image(x_fraud, f"FRAUD (idx={fraud_idx}, proba={p_fraud:.3f})")
img_norm  = waterfall_to_image(x_normal, f"NORMAL (idx={normal_idx}, proba={p_normal:.3f})")

# 7) 한 페이지(1행 2열)로 출력
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].imshow(img_fraud)
axes[0].axis("off")
axes[1].imshow(img_norm)
axes[1].axis("off")
plt.suptitle("SHAP Waterfall Comparison (Top5 RF): Fraud vs Normal", y=1.02)
plt.tight_layout()
plt.show()
```

## Fraud Detection Decision Logic  
### 1) 거래 금액 자체의 이상성 (amt)  
### 2) 최근 1일 소비 폭증 (cc_amt_sum_1d)  
### 3) 거래 빈도 변화 (cc_txn_cnt_1d)  
### 4) 오랜 공백 후 재등장 거래 (since_prev_txn_sec)   
### 5) 단기 폭발 결제 (cc_amt_sum_1h)
